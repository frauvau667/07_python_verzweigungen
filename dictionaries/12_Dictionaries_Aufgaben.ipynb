### Aufgabe 4.2: keys, values, items

Schreiben Sie hinter jede Zeile, was ausgegeben wird. Prüfen Sie dann durch Ausführen.

<details>
<summary><b>💡 Hinweis</b> (zum Aufklappen anklicken)</summary>

</details>

```python
hauptstadt = {"Frankreich": "Paris", "Italien": "Rom", "Spanien": "Madrid"}
print(len(hauptstadt))               # Vermutung:
print(list(hauptstadt.keys()))       # Vermutung:
print(list(hauptstadt.values()))     # Vermutung:
print("Italien" in hauptstadt)       # Vermutung:
print("Rom" in hauptstadt)           # Vermutung:
```
