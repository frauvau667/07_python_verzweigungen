## 4 Sichtbarkeit von Variablen

### Aufgabe 4.1: Global oder lokal?

Schreiben Sie hinter jede `print`-Zeile, was ausgegeben wird. Prüfen Sie dann durch Ausführen. Was passiert in der letzten Zeile?

<details>
<summary><b>💡 Hinweis 1</b> (zum Aufklappen anklicken)</summary>

Eine Zuweisung in der Funktion erzeugt eine **neue lokale** Variable – die globale bleibt unverändert.

</details>

<details>
<summary><b>💡 Hinweis 2</b> (zum Aufklappen anklicken)</summary>

Lokale Variablen existieren nach dem Funktionsende nicht mehr.

</details>

````python
x = 10

def test():
    x = 99
    y = 5
    print(f"innen: {x} {y}")   # Vermutung:

test()
print(f"außen: {x}")          # Vermutung:
print(y)                    # Vermutung:
````

### Aufgabe 4.2: Code aufräumen

Das Programm berechnet dreimal auf dieselbe Weise einen Bruttopreis inklusive 19 % Mehrwertsteuer und gibt ihn aus. Lagern Sie die Berechnung in eine Funktion mit Parameter und Rückgabewert aus, sodass die Formel nur noch **einmal** im Code steht.

<details>
<summary><b>💡 Hinweis 1</b> (zum Aufklappen anklicken)</summary>

Die sich wiederholende Rechnung `preis * 1.19` gehört in die Funktion; der Preis wird als Parameter übergeben.

</details>

<details>
<summary><b>💡 Hinweis 2</b> (zum Aufklappen anklicken)</summary>

Ändert sich der Steuersatz, muss man danach nur noch eine Stelle anpassen.

</details>

````python
preis1 = 49.90
brutto1 = round(preis1 * 1.19, 2)
print(f"Artikel 1: {brutto1} €")

preis2 = 12.50
brutto2 = round(preis2 * 1.19, 2)
print(f"Artikel 2: {brutto2} €")

preis3 = 199.00
brutto3 = round(preis3 * 1.19, 2)
print(f"Artikel 3: {brutto3} €")
````
